```python
import nltk
nltk.download('treebank')  # one-time; skips if already downloaded
from  nltk.corpus import treebank
```

```python
tagged_sents = treebank.tagged_sents()
parsed_sents = treebank.parsed_sents()
words = treebank.words()

print(f"Tagged sentences: {len(tagged_sents)}")
print(f"Parsed sentences: {len(parsed_sents)}")
print(f"Total tokens: {len(words)}")
```

```python
tags = sorted({tag for sent in tagged_sents for word, tag in sent})
print(f"Unique tags: {len(tags)}")
print(tags)
```

```python
examples = [w for sent in tagged_sents[:50] for w, t in sent if t == '-NONE-'] #Daily Mirror sentences will never have these, so they need stripping before training
print(examples[:15])
```
