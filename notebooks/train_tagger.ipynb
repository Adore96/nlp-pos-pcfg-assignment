```python
import sys, random
sys.path.insert(0, '..')
from src.data_utils import load_clean_treebank

clean_tagged_sents = load_clean_treebank()

random.seed(42)
shuffled = clean_tagged_sents[:]
random.shuffle(shuffled)
split = int(len(shuffled) * 0.9)
train_sents = shuffled[:split]
test_sents = shuffled[split:]
print(f"train: {len(train_sents)}, test: {len(test_sents)}")
```

```python
from nltk.tag import hmm

trainer = hmm.HiddenMarkovModelTrainer()
naive_tagger = trainer.train_supervised(train_sents)
print(naive_tagger.accuracy(test_sents))
```

```python
import sys
sys.path.insert(0, '..')
from src.tagger import lidstone_estimator
from nltk.tag import hmm

trainer = hmm.HiddenMarkovModelTrainer()
tagger = trainer.train_supervised(train_sents, estimator=lidstone_estimator)
print(tagger.accuracy(test_sents))
```

```python
import pickle

with open('../models/hmm_tagger.pkl', 'wb') as f:
    pickle.dump(tagger, f)
```
